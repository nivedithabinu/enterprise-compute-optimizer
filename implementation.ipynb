```python
import numpy as np 
import pandas as pd
```

```python
# 1. Loading the raw Kaggle data
load = pd.read_csv("data/apple_mac_performance_history.csv")
```

```python
# 2. Filter to the necessary columns and drop missing values
filteredColumns = ['model_name', 'geekbench6_single_core', 'geekbench6_multi_core']
load = load[filteredColumns].dropna()
```

```python
# 3. Engineer the synthetic market price using the correct column name
load['estimated_market_price'] = (300+(load['geekbench6_multi_core']*0.15)).round(2)
```

```python
# 4. Filter out performance anomalies (scores<100) and reset the index
load = load[load['geekbench6_multi_core']>1000].reset_index(drop=True)
```

```python
# 5. Export the clean, static dataset for the backend algorithm
load.to_csv('data/mac_performance.csv', index=False)
```
